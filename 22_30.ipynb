```python
import numpy as np
import pandas as pd

# ==========================================
# 22. Tạo và hiển thị DataFrame
# ==========================================
exam_data = {
    "name": [
        "Anastasia",
        "Dima",
        "Katherine",
        "James",
        "Emily",
        "Michael",
        "Matthew",
        "Laura",
        "Kevin",
        "Jonas",
    ],
    "score": [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    "attempts": [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    "qualify": ["yes", "no", "yes", "no", "no", "yes", "yes", "no", "no", "yes"],
}
labels = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"]

df = pd.DataFrame(exam_data, index=labels)
print("--- 22. DataFrame vừa tạo ---")
print(df)
print("\n" + "=" * 50 + "\n")

# ==========================================
# 23. Cho biết 3 dòng đầu tiên
# ==========================================
print("--- 23. 3 dòng đầu tiên ---")
print(df.head(3))
print("\n" + "=" * 50 + "\n")

# ==========================================
# 24. Chọn cột 'name' và 'score'
# ==========================================
print("--- 24. Chọn cột 'name' và 'score' ---")
print(df[["name", "score"]])
print("\n" + "=" * 50 + "\n")

# ==========================================
# 25. Chọn những dòng có attempts > 2
# ==========================================
print("--- 25. Dòng có attempts > 2 ---")
print(df[df["attempts"] > 2])
print("\n" + "=" * 50 + "\n")

# ==========================================
# 26. Cho biết số dòng, số cột của DataFrame
# ==========================================
print("--- 26. Số dòng và số cột ---")
so_dong, so_cot = df.shape
print(f"Số dòng: {so_dong}")
print(f"Số cột: {so_cot}")
print("\n" + "=" * 50 + "\n")

# ==========================================
# 27. Những dòng có score là NaN (Sử dụng isnull())
# ==========================================
print("--- 27. Dòng có score là NaN ---")
print(df[df["score"].isnull()])
print("\n" + "=" * 50 + "\n")

# ==========================================
# 28. Những dòng có score trong khoảng (15, 20) (Sử dụng between())
# Note: Hàm between mặc định bao gồm cả 2 đầu mút [15, 20]
# ==========================================
print("--- 28. Dòng có score trong khoảng (15, 20) ---")
print(df[df["score"].between(15, 20)])
print("\n" + "=" * 50 + "\n")

# ==========================================
# 29. Những dòng có attempt > 2 và score trong khoảng (15, 20)
# ==========================================
print("--- 29. Dòng có attempts > 2 VÀ score thuộc (15, 20) ---")
dieu_kien = (df["attempts"] > 2) & (df["score"].between(15, 20))
print(df[dieu_kien])
print("\n" + "=" * 50 + "\n")

# ==========================================
# 30. Thay đổi giá trị dòng 'd', cột score thành 19 (Sử dụng loc)
# ==========================================
print("--- 30. Thay đổi giá trị dòng 'd', cột 'score' thành 19 ---")
# Sử dụng .loc[tên_dòng, tên_cột] để thay đổi giá trị
df.loc["d", "score"] = 19

print("DataFrame sau khi thay đổi:")
print(df)
```
